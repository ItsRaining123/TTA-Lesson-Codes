```python
from ugot import ugot
import time

got = ugot.UGOT()
got.initialize("192.168.1.82")
got.show_light_rgb([1, 2, 3], 255, 255, 255)
```

## Task 1

```python
obstacle_dist = got.read_distance_data(21)

# Move 20cm/s forward if no obstacle within 10 cm
while obstacle_dist > 10:
    obstacle_dist = got.read_distance_data(21)
    got.mecanum_move_speed(0, 20)

got.mecanum_stop()
```

## Task 2

```python
obstacle_dist = got.read_distance_data(21)

while obstacle_dist > 20:
    obstacle_dist = got.read_distance_data(21)
    got.mecanum_turn_speed(2, 20)

got.mecanum_stop()
time.sleep(1)

while obstacle_dist > 5:
    obstacle_dist = got.read_distance_data(21)
    got.mecanum_move_speed(0, 20)
    
got.mecanum_stop()
```

## Extension Task

```python
import random

def random_rotation():
    got.mecanum_turn_speed_times(random.randint(2, 3), 45, 90, 2)

def move_forward():
    obstacle_dist = got.read_distance_data(21)

    while obstacle_dist > 10:
        obstacle_dist = got.read_distance_data(21)
        got.mecanum_move_speed(0, 20)

    got.mecanum_stop()
    time.sleep(1)

turn_counter = 0
while turn_counter < 5:
    move_forward()
    random_rotation()
    turn_counter += 1
    print(turn_counter)
```
