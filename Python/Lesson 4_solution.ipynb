## Combining Conditionals
In this section, you will learn how to:
- use the operators `and`, `or` and `not` in conditions
- write nested conditionals

In the previous lesson we have learned how to branch our code into different sections, and controlling when these sections of code should run. For example, we do this by checking whether we have enough money to buy a drink, and whether the user input matches with the expected password. What happens when the situation requires you to check <span style="color: red;">2 or more conditions</span>?

Lets take the following as an example:

````python
password = input("Enter your password: ")
money = 10

if password == "kittycat" and money >= 5:
    print("You can buy the drink!")
else:
    print("Sorry, you cannot buy the drink")
````

In the above example, we ask the user to input their password. After that, we need to check for 2 conditions:
- whether the user's password is correct
- whether the user has enough money

You can combine conditions with the logical operators `and` and `or`. The operator `and` specifies that **all** the given conditions must be true at the same time. The operator `or` specifies that **at least one** of the given conditions must be true.

The following example demonstrates the use of `or`, to check if a number is not between 5 to 8:

````python
number = int(input("Please type in a number: "))

if number < 5 or number > 8:
    print("The number is not within the range of 5 to 8")
````

The following truth table contains the behaviour of these operators in different situations:

| a | b | a and b | a or b |
|---|---|---|---|
| False | False | False | False |
| False | True | False | True |
| True | False | False | True |
| True | True | True | True |

Sometimes it is necessary to know if something is <span style="color: red;">not</span> true. The operator `not(!)` negates a condition:
| a | not a |
|---|---|
| F | T |
| T | F |

## Nested Conditionals
Nesting: a general term in programming that means <span style="color: red;">placing a block of code inside another</span>. For example, placing an `if`, `elif`, or `else` inside another conditional statement. Nesting is not limited to conditional statements and can include other commands like loops (e.g. having a `while` loop inside another `while` loop).

Conditional statements can also be nested within other conditional statements. For example, the following program checks whether a number is above zero, **and then** whether it is odd or even:

````python
number = int(input("Please type in a number: "))

if number > 0:
    if number % 2 == 0:
        print("The number is even")
    else:
        print("The number is odd")
else:
    print("The number is negative or zero")
````

Some examples of how this program behaves:

```
Please type in a number: 3
The number is odd
```

```
Please type in a number: 18
The number is even
```

```
Please type in a number: -4
The number is negative or zero
```

With nested conditional statements it is important to get the <span style="color: red;">indentations</span> right. Indentations determine which branches are linked together. For example, an `if` branch and an `else` branch with the same amount of whitespace are determined to be branches of the same conditional statement.

The same result can often be achieved using either nested conditional statements or conditions combined with logical operators. The example below is functionally no different from the example above, in the sense that it will print out the exactly same things with the same inputs:

````python
number = int(input("Please type in a number: "))

if number > 0 and number % 2 == 0:
    print("The number is even")
elif number > 0 and number % 2 != 0:
    print("The number is odd")
else:
    print("The number is negative or zero")
````

One code isn't necessarily better than the other, but in different situations one or the other way may seem more logical. In this particular example most people tend to find the first version with nesting to be more intuitive.

## UGOT Movement Control with Laser Sensor
In this section, you will learn how to:
- read the distance measured by the robot's laser sensor
- use conditions to decide whether the robot should move or stop
- use a loop to keep checking for an obstacle

````python
from ugot import ugot

got = ugot.UGOT()
got.initialize("192.168.1.201") # Replace with your robot's IP address
got.show_light_rgb([1, 2, 3], 255, 255, 255)
got.play_audio_tts("Connection successful")
````

So far we have controlled the robot by telling it how far to move. However, what happens if there is an <span style="color: red;">obstacle in its path</span>? 

Instead of choosing a fixed movement distance, we can use the ***laser sensor*** to measure the distance to an object in front of the robot. Our program can then use this information to decide whether the robot should continue moving.

To read the laser sensor, use `got.read_distance_data(id)`. Replace the id with 21.

````python
distance = got.read_distance_data(21)
print(f"Distance: {distance} cm")
````

Here is what happens:
- `got.read_distance_data(21)` reads the distance from port (2)(1)st sensor.
- The returned value is stored in `distance` variable.
- The distance is measured in centimeters.
- A value of -1 means no data was obtained. It does not mean that the path is clear!

Keep the robot stationary and place a large, flat object in front of the sensor. Run the code, move the object, and run it again.

Does the value increase or decrease when the object moves closer?

### Deciding whether to move

Let's choose a stopping distance of 20 cm. The robot should only move when:
- the sensor has returned a reading
- the distance is greater than 20 cm

````python
if distance == -1:
    print("Something is wrong with the sensor, please check again")
elif distance > 20:
    print("The robot can move forward")
else:
    print("The robot should stop")
````

Some examples of how the program behaves:
| Sensor reading | Decision | Reason |
|---|---|---|
| 50 | Move forward | obstacle is further than 20 cm |
| 21 | Move forward | obstacle is further than 20 cm |
| 20 | Stop | obstacle reached stopping distance |
| 12 | Stop | obstacle reached stopping distance
| -1 | Stop | No sensor data obtained |

### Moving while checking the sensor
Previously, we used `got.mecanum_move_speed_times()` to move for a chosen distance or duration. For this task, we will use this instead: `got.mecanum_move_speed(0, 20)`. This command will move the robot forward 20 cm/s. Since it does not specify a travel distance, the robot will continue moving forward <span style="color: red;">until a stop command</span> is sent.

To keep checking the distance while the robot moves, we place the sensor reading and conditions inside a `while True` loop.

````python
stop_distance = 20
move_speed = 20

while True:
    distance = got.read_distance_data(21)

    if distance > stop_distance:
        got.mecanum_move_speed(0, move_speed)
    else:
        got.mecanum_stop()

        if distance == -1:
            print("No sensor data, please check for damage")
        else:
            print(f"Obstacle detected at {distance} cm. Robot stopped")
        break
````

Let's look at how this works:
- Read: The robot gets a distance reading during every repetition of the loop.
- Decide: The program checks if there is at least a 20 cm path for the robot to move ahead. If not, it proceeds to the `else` branch and stops the robot
- Explain: The nested if-else prints the reason for stopping.
- Exit: break ends the loop, so the robot does not automatically start moving again.
- Repeat: If the stopping condition is not true, the robot moves forward, and checks again.

The robot may travel a little farther while readings and commands are being processed. The stopping distance is a trigger, so do not expect the robot to stop at exactly 20 cm. Test slowly and increase the stopping distance if needed.
