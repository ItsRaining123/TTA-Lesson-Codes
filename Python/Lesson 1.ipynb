## How to use this notebook
1. Open it in a Jupyter-compatible environment with a Python 3 kernel.
2. Jupyter Notebooks have 2 different cells available: **Code** cells for typing code, and **Markdown** cells to write text (like this one)
3. To run **Code** cells, press on the "Execute Cell" button on the left. 
4. For **Markdown** cells, press "Esc" to display the formatted text
5. You can run cells in any order

**Notebook memory:** Python remembers variables from earlier cells until its kernel is restarted. You do not need to run every cell everytime unless you made changes to it

**If something goes wrong:** read the last line of the error, locate the relevant code line, check spelling/quotes/colons/indentation, and change one thing. If a cell is stuck, use Interrupt/Stop. A restart clears remembered values but does not erase saved code. Save your work at the end of every lesson.

**Python virtual environment (venv):** A Python virtual environment (created via the built-in venv module) is an isolated directory that contains its own Python interpreter and independent set of installed **packages**. It prevents version conflicts between different projects and keeps your global Python installation clean.

**Package management:** Python packages are folders of reusable code files made by other developers that help you build programs without starting from scratch. Examples of these Python packages are "UGOT" and "OpenCV". Use `uv` to manage these your environment and packages. Some commands to know:
- ``uv init``: Creates the starting set of files and a clean Python virtual environment. Use this when you first create a new Python project
- ``uv add``: Installs a Python package in the current environment.
- ``uv remove``: Remove a Python package from the current environment. Use it if you accidently added a package or no longer need to use a specific Python package

## What is Python?
Python is a programming language—a way to give instructions to a computer. You can use it to create games, solve maths problems, work with data, and build useful tools.

Python was created by **Guido van Rossum** and first released in 1991. Its name came from the comedy group Monty Python, not the snake!

If you have used block-based coding, you have already practised arranging instructions. In Python, you **type** instructions instead of joining blocks. You will still use familiar ideas such as repeating actions and making decisions.

Python’s readable style makes it a popular language for beginners, but professional programmers use it too—even for data science and artificial intelligence.
