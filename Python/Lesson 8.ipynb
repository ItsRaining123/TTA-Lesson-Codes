## Lists and `for` loops
In this lesson, you will learn how to:

So far, each variable has held one value: one name, one score, one distance reading. But programs often need to work with a whole group of values. Making a separate variable for each one (`reading1`, `reading2`, `reading3`...) quickly becomes messy.

In this lesson, we'll learn how to keep a group of values together in a <span style="color: red;">list</span>, and how to use a `for` loop to work through every item in it.

## Creating a list
A list is written inside square brackets, `[]`, with a comma between each item. The items can be strings, numbers or Booleans, and you can print the whole list at once.

````python
fruits = ["apple", "banana", "cherry"]
scores = [72, 85, 90, 64]
empty = []

print(fruits)
print(scores)
print(empty)
````

Just like strings, we can use `len()` to find out how many items a list has. An empty list, `[]`, has a length of 0.

````python
fruits = ["apple", "banana", "cherry"]
print(len(fruits))
print(len([]))
````

## Indexing - Accessing items in a list
Lists work very much like the strings. Each item has an <span style="color: red;">index</span>, and Python <span style="color: red;">starts counting at 0</span>. Negative indexes count from the end of the list.

| Item | `"apple"` | `"banana"` | `"cherry"` |
|---|---|---|---|
| Index | 0 | 1 | 2 |
| Negative index | -3 | -2 | -1 |

````python
fruits = ["apple", "banana", "cherry"]
print("First item: " + fruits[0])
print("Second item: " + fruits[1])
print("Last item: " + fruits[-1])
````

Asking for an index that doesn't exist, such as `fruits[3]`, causes an `IndexError`, just like it did with strings.

Slices work on lists too. A slice of a list gives us a **new list** containing the selected items.

````python
scores = [72, 85, 90, 64, 58]
print(scores[:3])  # the first three scores
print(scores[-2:]) # the last two scores
````

## Lists can be changed

### Adding and removing items
Lists can also grow and shrink while the program runs. Here are some of the most useful list methods:

| Method | What it does |
|---|---|
| `my_list.append(item)` | Add `item` to the **end** of the list |
| `my_list.insert(index, item)` | Add `item` at the given `index`, moving later items along |
| `my_list.remove(item)` | Remove the **first** item equal to `item` |
| `my_list.pop()` | Remove the last item and give it back |
| `my_list.pop(index)` | Remove the item at `index` and give it back |

### Checking if an item is in a list
We can use `in` to check whether a list contains an item, just like we searched for text in a string. The result is a Boolean, so it fits straight into an `if` statement.

***Important note:*** `.remove()` causes a `ValueError` if the item isn't in the list. Checking with `in` first avoids that error.

````python
tasks = ["charge", "scan", "report"]
print("scan" in tasks)
print("dance" in tasks)

item = "dance"
if item in tasks:
    tasks.remove(item)
else:
    print(f"'{item}' is not in the list.")
````

## Looping through a list with `for`
We often want to do the same thing to every item in a list. Using what we already know, we could use a `while` loop and an index, just like we did with strings:

````python
fruits = ["apple", "banana", "cherry"]
index = 0
while index < len(fruits):
    print(f"I like {fruits[index]}")
    index += 1
````

### `for` loops work on strings too
A string is a sequence of characters, so a `for` loop can go through it one character at a time.

````python
for letter in "robot":
    print(letter)
````

### `while` or `for`?
- Use a **`for` loop** when you already know what to loop through: every item in a list, every character in a string, or a fixed number of times.
- Use a **`while` loop** when you keep going until a condition changes, and you don't know in advance how many times that will be. For example: keep asking until the user types something, or keep driving until the sensor sees an obstacle.

## Counting with `range()`
Sometimes we don't have a list but we want to repeat something a certain number of times. The `range()` function creates a sequence of numbers for a `for` loop to work through.

| Expression | Numbers produced |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(2, 6)` | 2, 3, 4, 5 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 |
| `range(5, 0, -1)` | 5, 4, 3, 2, 1 |

````python
for i in range(5):
    print(i)

print("---")
for i in range(2, 6):
    print(i)

print("---")
for i in range(0, 10, 2):
    print(i)
````

### Looping with an index using `range(len(...))`

````python
tasks = ["charge", "scan", "report"]
for i in range(len(tasks)):
    print(f"Task {i + 1}: {tasks[i]}")
````

## Using loops to work with lists
### Building a list in a loop
We can start with an empty list and `.append()` a new item each time through the loop. This is how a program collects values one by one, such as user answers or sensor readings.

````python
squares = []
for n in range(1, 6):
    squares.append(n * n)
print(squares)
````

### Built-in shortcuts
Totals, smallest and largest values are needed so often that Python has built-in functions for them: `sum()`, `min()` and `max()`. It's still important to understand the loops above, because you'll often need to do something these shortcuts can't, like counting only the passes.

***Note:*** `min()` and `max()` cause a `ValueError` on an empty list, and dividing by `len([])` causes a `ZeroDivisionError`. Check that a list isn't empty before using them.

````python
scores = [72, 85, 90, 64, 58]
print(sum(scores))
print(min(scores))
print(max(scores))
````

## UGOT Control
Lists and `for` loops let the robot work through a whole sequence of instructions, or collect a set of sensor readings, without repeating the same code over and over.

### Robot command reference
| Command | Meaning |
|---|---|
| `got.mecanum_move_speed_times(0, 10, distance, 1)` | Move forward at 10 cm/s for `distance` cm, then stop |
| `got.mecanum_turn_speed_times(2, 30, 90, 2)` | Turn left 90 degrees at 30 degrees/s |
| `got.mecanum_turn_speed_times(3, 30, 90, 2)` | Turn right 90 degrees at 30 degrees/s |
| `got.read_distance_data(21)` | Read the distance (in cm) from the laser sensor |
| `got.show_light_rgb([1, 2, 3], r, g, b)` | Set the lights to a colour using red, green and blue values from 0 to 255 |
| `got.play_audio_tts(text, wait=True)` | Say the text and wait for speech to finish |
| `got.screen_clear()` | Clear the previous screen content |
| `got.screen_display_background(0)` | Set a black background |
| `got.screen_print_text_newline(text, color)` | Display a string and move to the next line |
