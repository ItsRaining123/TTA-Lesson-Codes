## Loops - Debugging and helper variables
Let's make the password checking example from Lesson 3 Activity 6 a bit more realistic. This version gives the user only <span style="color: red;">three attempts</span> at typing in a password.

The program uses ***two helper variables***. The variable `attempts` keeps track of how many times the user has typed in a password. The variable `success` is set to either True or False based on whether the user is successful in signing in.

````python
attempts = 0

while True:
    password = input("Please type in your password: ")
    attempts += 1

    if password == "kittycat":
        success = True
        break

    if attempts == 3:
        success = False
        break

    # this is printed if the code was incorrect AND there have been less than three attempts
    print("Incorrect...try again")

if success:
    print("Correct PIN entered!")
else:
    print("Too many attempts. Exiting program...")
````

The loop is exited either when the user types the correct password OR if there have been too many attempts. The if statement after the loop checks the value of the variable `success` and prints out a message accordingly.

Adding loops to programs also adds to the potential sources of <span style="color: red;">bugs</span>. It becomes even more important to master the use of debugging print statements in your code.

***What are bugs in Python?***
- Bugs in Python, or programming in general, refers to an error, flaw, or mistake in the code that causes a program to produce unintended result, freeze, or crash

Let's have a look at a program almost identical to the previous example, but with one important difference:

````python
attempts = 0

while True:
    password = input("Please type in your password: ")
    attempts += 1

    if attempts == 3:
        success = False
        break

    if password == "kittycat":
        success = True
        break

    print("Incorrect...try again")

if success:
    print("Correct PIN entered!")
else:
    print("Too many attempts. Exiting program...")
````

This version acts strangely when the user types in the correct password on the third attempt. So, let's try and find the cause by adding some strategic debugging print statements inside the loop:

````python
attempts = 0

while True:
    password = input("Please type in your password: ")
    attempts += 1

    print(f"Attempt {attempts}")
    print(f"Condition 1: {attempts == 3}")
    if attempts == 3:
        success = False
        break

    print(f"Password: {password}")
    print(f"Condition 2: {password == "kittycat"}")
    if password == "kittycat":
        success = True
        break

    print("Incorrect...try again")

if success:
    print("Correct PIN entered!")
else:
    print("Too many attempts. Exiting program...")
````

From the above printouts we can see that during the ***third iteration of the loop*** the condition of the first `if` statement is `True`, and the loop is exited. This iteration never gets to the second `if` statement, which checks whether the code was typed in correctly.

The order of conditional statements, or different branches within a conditional statement, is a common cause for bugs, especially in loops. Debugging print statements are often the simplest way of finding their cause.

## Joining strings with the + operator
The above password checking example used a helper variable `attempts` to keep track of how many times the user had tried to type in a code. The variable is set to zero outside the loop, and each iteration increases its value by one. A similar idea of <span style="color: red;">incrementation works with string variables</span> as well. The program could, for instance, keep track of all the passwords the user typed in:

````python
attempts = 0
passwords = ""

while True:
    password = input("Please type in your password: ")
    attempts += 1
    passwords = passwords + f"{password}, "

    if password == "kittycat":
        success = True
        break

    if attempts == 3:
        success = False
        break

    print("Incorrect...try again")

if success:
    print("Correct PIN entered!")
else:
    print("Too many attempts. Exiting program...")

print(passwords)
````

The helper variable is initialized to an empty string. With each iteration the string gets longer, as the code the user typed in is added, along with a comma. If the user types in the passwords *abcd* *robot* *qwertyuiop*, at the end of the program's execution the value of `passwords` would be:

```
abcd, robot, qwertyuiop
```

## Loops with conditions
In this section, you will learn:
- how to create a `while` loop with a condition
- what roles initialisation, formulating a condition and updating variables does in a loop
- how to create loops with different kinds of conditions

In the previous few sections we learnt to use the `while True` loop to repeat sections of code. The condition of the loop is `True`, so the condition is fulfilled every time. We needed to explicitly <span style="color: red;">break out from the loop</span> each time to avoid an infinite loop. For example:

````python
# Print numbers until the variable a equals 5
a = 1

while True:
    print(a)
    a += 1

    if a == 5:
        break
````

Of course, the condition doesn't always have to be `True`, but instead <span style="color: red;">any Boolean expression</span> can be used as the condition. The general structure of the `while` statement is as follows:

```python
while <condition>:
    do something here...

other parts of program...
```

The idea here is that the execution goes back and forth, checking if the condition is true and executing the code within the block, over and over again. If the condition at any point is `False`, <span style="color: red;">execution of the program continues</span> from the line after the `while` block.

![while condition flowchart](./Lesson%205%20while%20condition.png)

Going back to the above example, if we use a while condition loop instead, we can create the equivalent as follows:

````python
a = 1

while a < 5:
    print(a)
    a += 1
````

In this structure the condition is always checked before the block within the loop is executed. It may happen that the block never gets executed, like so:

````python
a = 6 # Changed from 1 to 6

while a < 5:
    print(a)
    a += 1
````

### Initialization, condition and update
To create a loop you'll often need to include three distinct steps: ***initialization, condition, and updating the iteration variables***.
- Initialization refers to setting the starting value(s) of the variable(s) used within the condition of the loop. These are often called the iteration or iterator variables. This is performed before the loop is first entered. 
- The condition defines for how long the loop is to be executed. It is set out at the very beginning of the loop. 
- Within each repetition of the loop the variables involved in the condition are updated, so that each iteration brings the loop one step closer to its conclusion. 

The following example illustrates these steps:

````python
number = int(input("Please type in a number")) # Initialization

while number < 10: # Condition
    print(number)
    number += 1 # Updating iteration variables

print("Execution finished")
````

If any one of these three components is missing, the loop will likely NOT function correctly. <span style="color: red;">A typical error is omitting the update step</span>:
```python
number = 1

while number < 10:
    print(number)

print("Execution finished.")
```

Here, the value of the variable `number` never changes. The program is stuck in an infinite loop, and the exact same bit of code is repeated over and over again until the user stops the execution by pressing `Control` +  `C`

### Activity 5 - Robot Journey Recorder
Please write a program that allows the user to control the robot using a command menu.

The robot should accept the following commands:
| Command | Action |
|---|---|
| forward | Move forward 20 cm |
| left | Turn left 90 degrees |
| right | Turn right 90 degrees |
| backward | Move backward 20 cm |
| quit | End the program |

Your program should end when the robot has completed five movement commands, or when the user enters quit.

Use helper variables to keep track of:
- the number of movement commands completed
- the journey as a string
- whether the program should continue running

Only valid movement commands should count towards the five-command limit. An invalid command should display a message without changing the movement count or journey.

At the end, print the number of movements completed and the journey.

For example:
```

Moves completed: 1
Command (forward/left/right/backward/quit): jump
Command not recognised.
Command (forward/left/right/backward/quit): left
```

Hint: You can begin with these helper variables:
```

```

````python
# TODO: Write your answer here
moves = 0
journey = ""
running = True

while moves < 5 and running:
    command = input("Command (forward/left/right/quit): ")
    valid_move = False

    if command == "forward":
        got.mecanum_move_speed_times(0, 10, 20, 1)
        valid_move = True
    elif command == "backward":
        got.mecanum_move_speed_times(1, 10, 20, 1)
        valid_move = True
    elif command == "left":
        got.mecanum_turn_speed_times(2, 45, 90, 2)
        valid_move = True
    elif command == "right":
        got.mecanum_turn_speed_times(3, 45, 90, 2)
        valid_move = True
    elif command == "quit":
        running = False
    else:
        print("Command not recognised.")

    if valid_move:
        moves += 1
        journey += f"{command} "

        print(f"Moves completed: {moves}")

print("Robot journey finished.")
print(f"Moves completed: {moves}")
print(f"Journey: {journey}")
````
