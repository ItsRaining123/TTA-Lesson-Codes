## Joining and repeating strings
You used `+` in an earlier lesson to join a greeting to a name. Joining strings like this is called <span style="color: red;">concatenation</span>.

Python puts the strings together exactly as they are, so it won't add a space for you. Compare the two greetings below and notice where the space comes from.

````python
first = "Hello"
name = "Alex"
print(first + name)
print(f"{first} {name}!")
````

We can also repeat a string using `*`. Give it a string and an integer, and Python will repeat the string that many times.

````python
word = "banana"
print(word * 3)
````

What happens if we ask Python to repeat a string `0` times? We get an empty string, `""`, with no characters in it. A negative repetition count gives the same result.

**Note:** When joining text and numbers, remember that `"5"` is a string and `5` is an integer. You can turn the number into text with `str()` or include it in an f-string, as shown below.

````python
score = 5
print("Score: " + str(score))
print(f"Score: {score}")
print("Start" + "-" * 0 + "End")
print(f"Start{"-" * 0}End")
````

Using string operations together with a loop we can write a program which draws a pyramid:

````python
n = 10 # number of layers in the pyramid
row = "*"

while n > 0:
    print(" " * n + row)
    row += "**"
    n -= 1
````

The `print` command within the loop prints a line, which begins with `n` spaces, followed by whatever is stored in the variable `row`. Then two stars are added to the end of the variable `row`, and the value of the variable `n` is decreased by 1.

Some things to note when determining the length of the string:

## Indexing - Accessing individual characters
We can also pick out a single character from a string. Each character has a position, called an <span style="color: red;">index</span>, and Python <span style="color: red;">starts counting these positions at 0</span>.

To get a character, put its index inside square brackets, `[]`, after the string. The positions in ROBOT look like this:

| Character | R | O | B | O | T |
|---|---|---|---|---|---|
| Index | 0 | 1 | 2 | 3 | 4 |
| Negative index | -5 | -4 | -3 | -2 | -1 |

You can count from the other end using negative indexes. The last character has index `-1`, so you can access it without first working out the length of the string.

````python
word = "ROBOT"
print("First character: " + word[0])
print("Third character: " + word[2])
print("Last character: " + word[-1])
print("Second Last character: " + word[-2])
````

Because we start counting at zero, the last character is at `len(word) - 1`. Using `len(word)` as an index would take us one position past the end.

For example, `"ROBOT"` has five characters, with indexes from 0 to 4. Both lines below access the same final character.

````python
word = "ROBOT"
print(word[len(word) - 1])
print(word[-1])
````

### Avoiding an IndexError
If we ask for a character at a position that doesn't exist, Python gives us an error. For example, `"ROBOT"[5]` would cause an `IndexError` because its last character is at index 4.

An empty string has no characters at all, so even asking for its first character will fail. Read the example below without running it and think about what would happen:
```python
word = ""
print(word[0])
```

We can avoid this error by checking the length before accessing a character. Run the next example twice: once with a word, and once by pressing Enter without typing anything.

````python
word = input("Type a word: ")
if len(word) > 0:
    print(f"First character: {word[0]}")
else:
    print("There is no first character: the string is empty.")
````

## Looping Strings - Strings and while loops
Now that we can access a character by its index, we can use a loop to work through a whole string. Just like how your robot travels forward while the reading from the laser sensor is still within certain range:

```python
distance = got.read_distance_data(21)
stop_distance = 20

while distance > stop_distance:
    got.mecanum_move_speed(0, 20)
    distance = got.read_distance_data(21)
```

The index starts at 0 and increases by one each time through the loop. We stop when it reaches the length of the string.

As you read the example, think about why the condition uses `<` instead of `<=`. What would happen if the loop ran one more time?

````python
word = "robot"
index = 0
while index < len(word):
    print(f"Index {index}: {word[index]}")
    index += 1
````

## Substrings and slices
So far, we've picked out one character at a time. We can also take a whole section of a string. This is called a substring. A substring of a string is a sequence of characters that forms a part of the string. For example, `"bot"` is a substring of `"robot"`.

To extract a substring, we use a <span style="color: red;">slice</span>. With `text[start:end]`, Python begins at `start` and takes the characters up to, but not including, `end`.

| Expression | Result for `word = "PYTHON"` |
|---|---|
| `word[0:2]` | `"PY"` |
| `word[2:5]` | `"THO"` |
| `word[:3]` | `"PYT"` |
| `word[3:]` | `"HON"` |
| `word[-2:]` | `"ON"` |

You can leave out either end of the slice. If you leave out the start `[:3]`, Python begins at index 0. If you leave out the end `[3:]`, it takes the rest of the string.

What if the end of a slice goes beyond the string? Python takes the characters that are available, so this won't cause an `IndexError`. If the range contains no characters at all, the slice gives us an empty string. Try the examples below and compare the results.

````python
word = "robot"
print(word[:20])
print(f"Slice beyond the end: '{word[20:]}'")
print(f"Empty input sliced: '{''[:3]}'")
````

### Strings cannot be changed one character at a time
We can read individual characters, but we can't replace them by assigning to an index. Strings are immutable, which means their contents cannot be changed after they are created.

For example, `word[0] = "R"` would cause a `TypeError`. To change the first letter, we can build a new string from the letter we want and a slice of the original word, then store it in the same variable.

````python
word = "robot"
word = "R" + word[1:]
print(word)
````

## Searching for substrings
Sometimes we want to check whether a word or part of a word appears in a message. We can do this with `in`, which gives us a Boolean result: `True` or `False`.

For a match, the characters must appear together, in the same order. Searches are also case sensitive.

````python
message = "The robot is ready"
print("robot" in message)
print("Robot" in message)
print("bot" in message)
print("rbt" in message)
````

Since the search gives us a Boolean, we can use it directly in a condition. This program checks the message and chooses a response:

````python
message = input("Message: ")
if "robot" in message:
    print("You mentioned a robot!")
else:
    print("The text 'robot' was not found.")
````

### Finding the position with .find()
Once we know how to check for a match with `in`, we might also want to know where that match begins, or where exactly it is. We can use `.find()` to get the index of the first occurrence.

When it finds the text, `.find()` returns its starting index, including 0 if the match is at the beginning. When there is no match, it returns -1.

````python
word = "banana"
print(word.find("ban"))
print(word.find("ana"))
print(word.find("z"))
````

***Important note:*** When we use the result in a condition, we need to check `position >= 0`. That includes a match at the very beginning of the string. Using `if position:` would give us the wrong result: Python treats 0 as false and -1 as true in a condition.

There's also a special case to watch for. An empty search string matches every string at index 0. In the next activity, we'll ask the user to enter a search term and check that they haven't left it empty.

````python
word = "banana"
print(word.find(""))
````

### Reading a text command
A command such as `"move forward"` gives us two pieces of information: the action, `"move"`, and the direction, `"forward"`. We can separate them by finding the space between the words.

Check that a space was found before using its index to slice the command. Read the example below and follow how the two slices separate the words.

````python
command = "move forward"
space = command.find(" ")
if space >= 0:
    action = command[:space]
    direction = command[space + 1:]
    print(f"Action: {action}")
    print(f"Direction: {direction}")
else:
    print("The command needs two words.")
````
