## Conditional Statements
In this section, you will learn how to:
- use a simple conditional statement in programming
- express conditionals with comparison operators

So far, every program we have written has been executed line by line in order. That works for printing a greeting, but real tasks need to <span style="color: red;">respond to changing data</span>. For this section, we are going to create sections of programs that will only execute in certain situations.

For example, the following program checks whether the user has enough money to buy a drink:

````python
money = int(input("How much money do you have? "))

if money > 3:
    print("You have enough money!")
    print("Here's your drink! Enjoy!")

print("Next customer please!")
````

When the user has over $3, the output of the program should look like this:

```
How much money do you have? 4
You have enough money!
Here's your drink! Enjoy!
Next customer please!
```

When the user has less than $3, the program output is only:
```
How much money do you have? 2
Next customer please!
```

These examples show us that the value given as input affects which parts of the program are executed. The program contains a <span style="color: red;">*conditional statement*</span> with a block of code which is executed only if the condition in the statement is **true**.


In a conditional statement the keyword `if` is followed by a <span style="color: red;">*condition*</span>, such as a comparison of two values. Conditional statements produce **Boolean values**, which the program use to make decisions. The code block following is only executed if the condition / Boolean value is true.

The resulting Boolean values from conditional statements into can be stored into variables just like the result from an arithmetic calculation:

````python
money = 5
is_enough = money > 3

print(is_enough)

if is_enough:
    print("You have enough money!")
````

Notice the colon character following the `if` header. In the following code there is no colon, run and check the output:

````python
money = 5

if money > 3
    print("You have enough money!")
````

### Comparison Operators
Very typically conditions consist of comparing two values. Here is a table with the most common comparison operators used in Python
| Operator | Purpose | Examples |
|---|---|---|
| `==` | Equal to | `a == b` <br> `gender == "Male"` |
| `!=` | Not equal to | `a != b` <br> `age != 15` |
| `>` | Greater than | `a > b` <br> `age > 15` |
| `>=` | Greater than or equal to | `a >= b` <br> `score >= 50` |
| `<` | Less than | `a < b` <br> `temperature < 35` |
| `<=` | Less than or equal to | `a <= b` <br> `temperature <= 35`|

### Difference between '=' and '=='
In programming, the difference between a single equal sign (=) and double equal sign (==) is that single equal sign is for **assignment**, while double equal sign is for **comparison**

| Operator | Purpose | Example | Result |
|---|---|---|---|
| `=` | Assigns a variable a value | x = 5 | The variable `x` now holds the value `5` |
| `==` | Checks if the values on the left and right are equal | x == 5 | Evaluates to either `True` or `False` depending on the value of `x` |

## Indentation

Python uses indentation to indicate a <span style="color: red;">block of code</span>, such as a part of a conditional statement, if each line of code in the block is indented the same. Meaning that each line in that block have the same amount of whitespace.

You can use the **Tab** key, to insert a set amount of whitespace. Many text editors like Visual Studio Code will automatically indent the following line when the **Enter** key is pressed after a colon character. When you want to end an indented code block you can use the **Backspace** key to return to the beginning of the line.

## Conditional Statements (Part 2)
In this section, you will learn how to:
- create multiple branches within conditional statements
- understand the purpose of `if`, `elif` and `else` statements within a conditional statement

Look at your code for Activity 1, what do you notice?
```python
number = int(input("Please type in a number: "))

if number > 0:
    print("The number is positive")

if number < 0:
    print("The number is negative")

if number == 0:
    print("The number is zero")
```

The code looks a bit clumsy and repetitive, because only one of the `if` blocks will execute (numbers can't be positive and negative at the same time). So, we only need to check if the number is negative. 

Instead of creating a whole another conditional statement, as in the example above, it is possible to create <span style="color: red;">another branch</span> of the same conditional statement to cover all cases where the condition was false. This is called the `else` statement.

The above code rewritten:

````python
number = int(input("Please type in a number: "))

if number < 0:
    print("The number is negative")
else:
    print("The number is positive or zero")
````

Note: there can never be an `else` branch without an `if` branch before it. The `if-else` construction as a whole forms a single conditional statement.

The following example checks whether a number given by the user is even or not. Parity can be checked with the modulo operator `%`, which produces the remainder of an integer division operation. When divided by two, if the remainder is zero, the number is even. Otherwise the number is odd.

````python
number = int(input("Please type in a number: "))

if number % 2 == 0:
    print("The number is even")
else:
    print("The number is odd")
````

Another example with string comparison:

````python
correct = "abcde"
password = input("Please type in the password: ")

if password == correct:
    print("Welcome")
else:
    print("No admittance")
````

### Alternative branches using `elif`
Often there are more than two options the program should account for. For example, the result of a football match could go three ways: home wins, away wins, or there is a tie.

A conditional statement can be added to with an `elif` branch. It is short for the words <span style="color: red;">"else if"</span>, which means the branch will contain an alternative to the original condition. Importantly, an `elif` statement is executed only if none of the branches before it is executed.

````python
goals_home = int(input("Home goals scored: "))
goals_away = int(input("Away goals scored: "))

if goals_home > goals_away:
    print("The home team won!")
elif goals_away > goals_home:
    print("The away team won!")
else:
    print("It's a tie!")
````

This program could print out three different statements given different inputs:

```
Home goals scored: 4
Away goals scored: 2
The home team won!
```

```
Home goals scored: 0
Away goals scored: 6
The away team won!
```

```
Home goals scored: 3
Away goals scored: 3
It's a tie!
```

There is no limit to the number of `elif` branches a conditional statement can contain, and the `else` branch is not mandatory.

````python
print("Holiday calendar")
date = input("What is the date today? ")

if date == "Dec 25":
    print("It's Christmas Day")
elif date == "Dec 31":
    print("It's New Year's Eve")
elif date == "Jan 1":
    print("It's New Year's Day")

print("Thanks and bye.")
````

Notice the example has no `else` branch. If the user inputs a date which is <span style="color: red;">not mentioned in any of the `if` or `elif` branches, or inputs a date in a different format</span>, none of the three branches of the conditional statement is executed.

## Loops
In this section, you will learn:
- what a loop means in programming
- to use a `while True` loop in your programs
- how to use the `break` command to break out of a loop

We have now learnt conditional structures in some detail. Another central technique in programming is repetition, or *iteration*. Together these form the fundamental <span style="color: red;">**control structures**</span> any programmer must master. They are called control structures because  they allow you to control which lines of code get executed and when.

While conditional structures allow you to choose between sections of code, iteration structures allow you to repeat sections of code. They are often called <span style="color: red;">loops</span> because they allow the program to "loop back" to some line that was <span style="color: red;">already executed before</span>. The process of executing one repetition of a loop is also referred to as an iteration of the loop.

This section introduces a simple `while` loop. Its structure is similar to the conditional statements we already covered. Let's have a look at a program which asks the user to type in a number and then prints out the number multiplied by 2. This continues until the user types in -1.

````python
while True:
    number = int(input("Please type in a number, -1 to quit: "))

    if number == -1:
        break

    print(number * 2)

print("Thanks and bye!")
````

As you can see above, the program asks for several numbers, thanks to the `while` statement in the program. When the user types in -1, the `break` command is executed, which <span style="color: red;">exits the loop</span> and execution continues from the first line <span style="color: red;">after</span> the `while` block.

With loops, it is crucial that there is always a way to exit the loop at some point in the code, otherwise the repetition could go on forever. To illustrate this, let's change the above example a little:

````python
number = int(input("Please type in a number, -1 to quit: "))

while True:
    if number == -1:
        break

    print(number * 2)

print("Thanks and bye!")
````

In this version the program asks the user to type in a number outside the loop. If the user types in any other number than -1, the loop is never exited from. This forms an **infinite loop**, which means the block of code within the loop is repeated endlessly.

The following program has a similar structure to the example above the infinite loop, but the user experience is quite different. This program allows the user to proceed **only** if they type in the correct PIN 1234:

````python
while True:
    code = input("Please type in your PIN: ")
    if code == "1234":
        break
    print("Incorrect...try again")

print("Correct PIN entered!")
````

## UGOT Movement Control with Conditional and Loops
In this section, we will learn how to control and decide whether the UGOT robot should move or not move using the control structures we have learned in this lesson.

````python
from ugot import ugot

got = ugot.UGOT()
got.initialize("192.168.1.201")
got.show_light_rgb([1, 2, 3], 255, 255, 255)
got.play_audio_tts("Connection successful")
````

Last lesson you have tried using user inputs to control the movement of the robot. Let's try to use `if-else` to achieve the same. First let's control if the robot should move forwards or backwards:

````python
direction = input("Select a direction (forward/backward): ")

if direction == "forward":
    got.mecanum_move_speed_times(0, 20, 50, 1)
elif direction == "backward":
    got.mecanum_move_speed_times(1, 20, 50, 1)
else:
    print("Input invalid")
````

Now do the same for turning movement:

````python
direction = input("Select a direction (left/right): ")

if direction == "left":
    got.mecanum_turn_speed_times(2, 45, 90, 2)
elif direction == "right":
    got.mecanum_turn_speed_times(3, 45, 90, 2)
else:
    print("Input invalid")
````
