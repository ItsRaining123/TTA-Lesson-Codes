## Data Types
In programming, data type is an important concept.

It allows you to check whether certain actions are valid or even allowed in the first place.

Some examples of data types in Python include:
| Type | Name | Meaning | Examples |
|---|---|---|---|
| `str` | String | Text, written in quotes | `"Nova"`, `"12"` |
| `int` | Integer | Numbers with no decimals or fractions | `12`, `0`, `-3` |
| `float` | Float | Number represented with floating-point precision | `2.5`, `12.0` |
| `bool` | Boolean | Truth value | `True`, `False` |

You can get the data type of any object by using the `type()` function:

````python
print(type("abc"))
print(type(5))
print(type(2.5))
print(type(True))
````

## User Input
The Python command `input`, reads a line of input typed in by the user. You can use this input to display a message to the user, or use it for some other operation.

The following program asks the user's name with the `input` command. Then prints it out with the `print` command.

````python
my_name = input("What is your name?")
print("Hi! " + my_name)
````

The execution of this program could look like this (input from the user in red):

What is your name? <span style="color: red;">John Python</span><br>

What this program prints out is partially dependent on input from the user. That means the execution of the program could also look like this:

`my_name` in the above program is called a *variable*. A *variable* is a location to hold some value, such as a string or number. This value can be used later, and it can also be changed

### More than one input
A program can ask for more than one input. Notice how below each `input` command stores the received value in a different variable.

````python
name = input("What is your name? ")
email = input("What is your email address? ")
birthday = input("When is your birthday? ")

print("Please double check your records:")
print("Your name: " + name)
print("Your email address: " + email)
print("Your birthday: " + birthday)
````

As implied by the name variable, the value stored in a variable can change. If the same variable is used to store more than one input, each new value will replace the previous one. For example:

````python
address = input("What is your address? ")
print("So you live at address " + address)

address = input("Please type in a new address: ")
print("Your address is now " + address)
````

This means that if the same variable is used to store two inputs in succession, there is no way to access the first input value after it has been replaced by the second:

````python
address = input("What is your address? ")
address = input("Please type in a new address: ")

print("Your address is now " + address)
````

### Printing with f-strings
You can also print using **string formatting (f-string)**. An **f-string** starts with `f` before the opening quote. Put a variable or expression inside `{}`
to insert its value. This lets you mix text and numbers without converting each number to text yourself.

*Note: F-strings were introduced in Python version 3.6. If you are using an older version of Python, f-strings may not work.*

````python
print(f"Hi {my_name}! Let me make sure: your name is {my_name}?")
````

A single f-string can contain multiple variables. For example this code

````python
name = "John"
city = "Singapore"
address = "123 ABC Street"
print(f"Hi {name}. You live in {city} at {address}")
````

It is difficult to create a printout exactly like this using the comma notation in the print command. For example, this program

````python
name = "John"
age = 22
city = "Singapore"
print("Hi", name, ". You live in", city, "at ", address)
````

The syntax can initially look a bit confusing, but in the end f-strings are often the simplest way of formatting text. This is actually the preferred way to format strings in Python nowadays, but feel free to use any methods based on your preferrence.

## Variables (Continued)
So far we have only stored strings in variables, but there are also many other types of information we will want to store and access later. Let's look at integers first

The following program creates the variable `age`, which contains an integer value

````python
age = 20
print(age)
````

Notice how we don't add quotation marks around the variable. In fact, if we were to add quotation marks around the number, this would mean our variable would no longer be an integer, but a string instead. This is similar to printing numbers like what we did on the previous lesson.

So, why does it matter that variables have a type, when the following program still prints out the same thing twice?
- Variable types matter because different operations affect different types of variables in different ways. Let's have a look at an example:

````python
number1 = 100
number2 = "100"

print(number1 + number1)
print(number2 + number2)
````

For integer values the + operator means addition, but for string values it means <span style="color: red;">concatenation</span>, or "stringing together".

Not all operators are available for all types of variables. While numbers can be divided using the division operator /, attempting to divide a string by a number causes an error:

````python
number = "100"
print(number / 2)
````

When printing floats or numbers with decimals (e.g. 2.45, 4.87...), you can specify how many decimal places to print by adding a colon (:) followed by a legal formatting type, e.g. `.2f` means 2 decimal places.

````python
price = 100 / 3
print(f"The price is: {price}")
print(f"The price is: {price:.2f}")
````

## Arithmetic Operations
Operators are used to perform operations on variables and values. In the following table you can see the most common arithmetic operators in Python, with examples:

| Operator | Purpose | Example | Result |
|---|---|---|---|
| `+` | Addition | 2 + 4 | 6 |
| `-` | Subtraction | 10 - 2.5 | 7.5 |
| `*` | Multiplication | 2 * 3 | 6 |
| `/` | Division | 9 / 2 | 4.5 |
| `//` | Integer Division | 9 // 2 | 4 |
| `%` | Modulo (Remainder) | 9 % 2 | 1 |

The order of operations is familiar from mathematics: BODMAS. The order can be changed with parentheses.

````python
print(2 + 3 * 3)
print((2 + 3) * 3)
````

- If two integers are added together, the result will also be an integer. 
- If a floating point number is subtracted from another floating point number, the result is a floating point number. 

In fact, if one of the numbers in the math expression is a floating point number, the result will also be a floating point number, regardless of the other numbers. 

Division / is an exception to this rule. Its result is a floating point number, even if the operands are integers. For example 1 / 5 will result in the floating point number 0.2.

Notice Python also has an integer division operator //. If the operands are integers, it will produce an integer. The result is <span style="color: red;">rounded down</span> to the nearest integer. For example this program

````python
x = 3
y = 2

print(f"/ operator {x/y}")
print(f"// operator {x//y}")
````

### Numbers as input
We have already used the `input` command to read in strings from the user. The same function can be used to read in numbers, but the string produced by the function must then be <span style="color: red;">converted to a numeric data type</span> in the program code.

A string can be converted into an integer with the function `int`. The following program asks the user for their year of birth and stores it in the variable input_str. The program then creates another variable `year`, which contains the year of birth stored in `input_str` converted into an integer. After this the calculation `2026-year` is possible, using the user-supplied value.

````python
input_str = input("Which year were you born? ")
year = int(input_str)
print(f"Your age at the end of the year 2026: {2026 - year}")
````

Usually you do not need to create two separate variables (like `input_str` and `year` above) to read a number value from the user. Instead, reading the input with the `input` function and converting it with the `int` function can be achieved in one go:

````python
year = int(input("Which year were you born? "))
print(f"Your age at the end of the year 2026: {2026 - year}")
````

Likewise, you can also convert integers into strings, like so:

````python
x = str(9)
print(type(x))
````

## Combining arithmetic operations with variables
Let's have a look at a program which calculates the sum of three numbers given by the user:

````python
number1 = int(input("First number: "))
number2 = int(input("Second number: "))
number3 = int(input("Third number: "))

sum = number1 + number2 + number3
print(f"The sum of the numbers: {sum}")
````

The program uses four different variables, but two would easily suffice in this case:

````python
sum = 0

number = int(input("First number: "))
sum = sum + number

number = int(input("Second number: "))
sum = sum + number

number = int(input("Third number: "))
sum = sum + number

print(f"The sum of the numbers: {sum}")
````

Now all inputs from the user are read into the one and the same variable number. The value of the variable sum is increased by the value of the variable number each time the user inputs a new number.

Increasing the value of a variable is a very common operation. As such, there is a commonly used shorthand notation which achieves the same result as the explicit summing up above:

```python
sum += number
```

````python
sum = 0

number = int(input("First number: "))
sum += number

number = int(input("Second number: "))
sum += number

number = int(input("Third number: "))
sum += number

print(f"The sum of the numbers: {sum}")
````

In fact, we don't necessarily need the variable number at all. The inputs from the user can also be processed like this:

````python
sum = 0

sum += int(input("First number: "))
sum += int(input("Second number: "))
sum += int(input("Third number: "))

print(f"The sum of the numbers: {sum}")
````

## UGOT Movement Control
In this section, we will learn how to control the movement of the UGOT robot vehicle using Python.

````python
from ugot import ugot

got = ugot.UGOT()
got.initialize("192.168.1.201")
got.show_light_rgb([1, 2, 3], 255, 255, 255)
got.play_audio_tts("Connection successful")
````

The UGOT mecanum wheel vehicle can perform two main types of movement:

1. **Move forward or backward**
2. **Turn left or right**

To control the robot, we need to provide different **inputs** to the movement actions. These inputs tell the robot things such as:
- Which direction to move
- How fast to move
- How far or how long to move
- Whether the movement is measured using distance, angle, or time

For example, the following command:

````python
got.mecanum_move_speed_times(0, 20, 50, 1)
````

tells the robot to:
- Move forward (0)
- At a speed of 20 cm/s
- Move for 50 cm
- Use centimeters as the movement unit (1)

Reference from the documentation:
```
mecanum_move_speed_times(direction, speed, times, unit)
Controls the mecanum wheel vehicle to move forward/backward for x seconds/cm and then stop.

Parameters
- direction (int) – Direction (0: Forward; 1: Backward)
- speed (int) – [5-80] Speed in centimeters per second
- times (int) – [0-360] Duration or range
- unit (int) – Unit type (0: Move in seconds; 1: Move in centimeters)

Returns
- None
```

Similarly:

````python
got.mecanum_turn_speed_times(3, 30, 90, 2)
````

tells the robot to:
- Turn right (3)
- At a speed of 30 degrees per second
- Turn by 90 degrees
- Use degrees as the turning unit (2)

By changing the values passed into these functions, we can control exactly how the robot moves.

Reference from the documentation:
```
mecanum_turn_speed_times(turn, speed, times, unit)<br>
Control the mecanum wheel vehicle to rotate left/right for x seconds/degrees and then stop.

Parameters
- turn (int) – Direction (2: Left turn; 3: Right turn)
- speed (int) – [5-280] Speed in degrees per second
- times (int) – [0-360] Duration or range
- unit (int) – Unit type (0: Move in seconds; 2: Move in degrees)

Returns
- None
```
