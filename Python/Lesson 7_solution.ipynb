## Parameters and arguments
The `ready_message` and `introduce_yourself` function always prints the same messages. Suppose we want to greet different people. We could write a separate function for each person, but the greeting instructions would be <span style="color: red;">almost identical</span>. For example:

```python
def introduce_alex():
    print("My name is Alex.")
    print("I enjoy building robots.")

def introduce_david():
    print("My name is David.")
    print("I like reading books.")

```

Instead, we can give the function the name it should use each time we call it. Functions often take one or more arguments, which may affect what the function does. Let's define some functions that take arguments. In the function definition, the parameters are defined within the parentheses after the function name:

***Note***: A **parameter** is a <span style="color: red;">variable named in a function's header</span> to receive information for a call. An **argument** is the value <span style="color: red;">we supply</span> when making that call. The terms parameter and argument can be used for the same thing: information that are passed into a function.

````python
def introduce_someone(name, hobby): # name and hobby are parameters
    print(f"My name is {name}")
    print(f"I like to {hobby}")
````

We can call it twice, with different information:

````python
introduce_someone("Jack", "play soccer") # "Jack" and "play soccer" are arguments
introduce_someone("Mary", "go shopping") # "Mary" and "go shopping" are arguments
````

On the first line, in the function header, we defined that this function takes in two arguments, and assigns it to a parameter named `name` and `hobby`. In the body of the function the `print` command uses the value stored in `name` and `hobby`.

Some things to note:
- By default, a function must be called with the correct number of arguments. A function that needs one argument **must** receive one. `introduce_someone()` or `introduce_someone("Alex")` would cause a `TypeError`.
- The order in which you supply the arguments matters because their positions determine which parameters receive them. Python does not read the meaning of the words and decide where they belong. If we swap the arguments, we change the values assigned to the parameters. Make sure your arguments are sent in the correct order

The names of functions and their parameters follow the same principles as the names of variables. They should be descriptive, and contain primarily lowercase letters and underscore characters. Again, there are some exceptions to these guidelines, but we will ignore those for now.

## Conditions and loops inside functions
The instructions you have used in previous lessons like `if-else`, `while` can also be written inside a function.

In the example below, `message` holds the text to print and `times` holds the number of repetitions. The `if` statement first checks whether `times` is negative. If it is, the function prints an explanation and skips the loop. Otherwise, the `else` block sets `count` to zero and starts checking the loop condition.

````python
def repeat_message(message, times):
    if times < 0:
        print("The number of repeats cannot be negative.")
    else:
        count = 0
        while count < times:
            print(message)
            count += 1

repeat_message("Ready", 2)
repeat_message("Go", 0)
repeat_message("Stop", -1)
````

Write a function `countdown(seconds)` for a resuable countdown in which the timer can be customized each time.
- If `seconds` is negative, print `Please use zero or a positive number.`
- Otherwise, use a `while` loop to print every number from `start` down to `0`, then print `Go!`.

Example outputs:

For `countdown(3)`:

For `countdown(-5)`:


For `countdown(0)`:
```text

```

````python
# TODO: Write your answer here
import time

def countdown(seconds):
    if seconds < 0:
        print("Please use zero or a positive number.")
    else:        
        while seconds > 0:
            print(seconds)
            time.sleep(1)
            seconds -= 1
        
        print("Go!")

countdown(0)
````

## Local variables and accidental globals
When a function runs, it can use its parameters and create helper variables inside its body. These are called **local variables** because they belong to that function call. In our earlier `sum` function, for example, `number1` and `number2` are parameters, while `total` is a helper variable created to hold their sum.

A variable assigned outside a function is called a **global variable**. A function can read a global variable, but doing so can make its behaviour depend on values that are not obvious from its arguments. Using global variables in a function is **usually a bad idea** because it makes the code <span style="color: red;">hard to read and understand</span>

Below is an example of a function which uses a global variable "by mistake":

````python
number = 10 # this is a global variable

def show_double(given_number):
    total = number * 2 # using the global variable instead of the "given_number" parameter 
    print(total)

show_double(3)
show_double(4)
````

No matter how many different arguments we call the function with, it will always print out the value "20" stored in the global variable "number".

## UGOT Control
In earlier lessons, you wrote robot commands directly in your program. We can now put related commands inside functions and call them whenever we need that action. For example, a greeting can include setting the lights, building a message and speaking it. Giving these instructions a function name lets us repeat the complete greeting with one call.

### Robot command reference
| Command | Meaning |
|---|---|
| `got.play_audio_tts(text, wait=True)` | Say the text and wait for speech to finish |
| `got.mecanum_move_speed_times(0, 10, distance, 1)` | Move forward at 10 cm/s for the supplied distance (in cm), then stop |
| `got.mecanum_stop()` | Stop the mecanum wheels |

````python
from ugot import ugot

got = ugot.UGOT()
got.initialize("192.168.1.170")  # Replace with your robot's IP address.
got.show_light_rgb([1, 2, 3], 255, 255, 255)
got.play_audio_tts("Connection successful")
````

### Activity 5 - Move and stop at an obstacle
Write a function called `move_stop(stopping_distance)` that moves the robot forward until the laser sensor detects an obstacle at or closer than the supplied stopping distance. The parameter `stopping_distance` must be greater than zero.

````python
# TODO: Write your answer here
def move_stop(stopping_distance):
    if stopping_distance <= 0:
        raise ValueError("Stopping distance must be greater than zero.")

    distance = got.read_distance_data(21)
    
    while distance > stopping_distance:
        got.mecanum_move_speed(0, 10)
        distance = got.read_distance_data(21)
    
    got.mecanum_stop()

move_stop(30)
````
