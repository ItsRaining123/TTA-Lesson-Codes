## Loops - Debugging and helper variables
Let's make the password checking example from Lesson 3 Activity 6 a bit more realistic. This version gives the user only <span style="color: red;">three attempts</span> at typing in a password.

The program uses ***two helper variables***. The variable `attempts` keeps track of how many times the user has typed in a password. The variable `success` is set to either True or False based on whether the user is successful in signing in.

````python
attempts = 0

while True:
    password = input("Please type in your password: ")
    attempts += 1

    if password == "kittycat":
        success = True
        break

    if attempts == 3:
        success = False
        break

    # this is printed if the code was incorrect AND there have been less than three attempts
    print("Incorrect...try again")

if success:
    print("Correct PIN entered!")
else:
    print("Too many attempts. Exiting program...")
````

## Loops with conditions
In this section, you will learn:
- how to create a `while` loop with a condition
- what roles initialisation, formulating a condition and updating variables does in a loop
- how to create loops with different kinds of conditions

In the previous few sections we learnt to use the `while True` loop to repeat sections of code. The condition of the loop is `True`, so the condition is fulfilled every time. We needed to explicitly <span style="color: red;">break out from the loop</span> each time to avoid an infinite loop. For example:

````python
# Print numbers until the variable a equals 5
a = 1

while True:
    print(a)
    a += 1

    if a == 5:
        break
````

Of course, the condition doesn't always have to be `True`, but instead <span style="color: red;">any Boolean expression</span> can be used as the condition. The general structure of the `while` statement is as follows:

```python
while <condition>:
    do something here...

other parts of program...
```

### Initialization, condition and update
To create a loop you'll often need to include three distinct steps: ***initialization, condition, and updating the iteration variables***.
- Initialization refers to setting the starting value(s) of the variable(s) used within the condition of the loop. These are often called the iteration or iterator variables. This is performed before the loop is first entered. 
- The condition defines for how long the loop is to be executed. It is set out at the very beginning of the loop. 
- Within each repetition of the loop the variables involved in the condition are updated, so that each iteration brings the loop one step closer to its conclusion.
